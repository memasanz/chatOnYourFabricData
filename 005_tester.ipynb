## Using a Service Principal with Fabric

```python
import os 
from sqlalchemy import create_engine
import pandas as pd
import time
from datetime import datetime

CLIENT_ID =  os.getenv("AZURE_CLIENT_ID")
CLIENT_SECRET = os.getenv("AZURE_CLIENT_SECRET")
server = os.getenv("FABRIC_CONNECTION_STRING")
database = os.getenv("FABRIC_WAREHOUSE")
driver = "ODBC Driver 18 for SQL Server"

connection_string = (f"mssql+pyodbc://{CLIENT_ID}:{CLIENT_SECRET}@{server}:1433/{database}"
 f"?driver={driver}"
 "&authentication=ActiveDirectoryServicePrincipal"
 "&timeout=120"
 "&Encrypt=yes"
 "&TrustServerCertificate=no"
) 

engine2 = create_engine(connection_string, connect_args={'timeout': 120})
query = "Select * from " + database + ".dbo." + os.getenv("FABRIC_TEST_TABLE")

print(query)
df = pd.read_sql_query(query, engine2)
print(df)
```
